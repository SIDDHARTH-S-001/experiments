```python
import torch
import torch.nn
```

```python
max_sequence_length = 2
d_model = 2
```

```python
even_i = torch.arange(0, d_model, 2).float()
even_i
```

```python
even_denominator = torch.pow(10000, even_i / d_model) # equation re-written for simplicity
even_denominator
```

```python
odd_i = torch.arange(1,d_model, 2).float()
odd_i
```

```python
odd_denominator = torch.pow(10000, (odd_i - 1) / d_model) # equation re-written for simplicity
odd_denominator # notice that its the same as even case
```

```python
denominator = even_denominator # hence writing a single variable
```

```python
# reshaping the positions to a 2D vector
position = torch.arange(max_sequence_length, dtype=torch.float).reshape(max_sequence_length, 1)
position
```

```python
even_PE = torch.sin(position / denominator)
odd_PE = torch.cos(position / denominator)
```

```python
print(even_PE)
print(odd_PE)
```

Interleave even and odd PE

```python
stacked = torch.stack([even_PE, odd_PE], dim=2)
print(stacked.shape)
PE = torch.flatten(stacked, start_dim=2)
print(PE.shape)
```
